# Lecture 11.1: Open Systems - Quantum Master Equation

## 1) Introduction to Open Systems

Consider a "two-level atom" with states $|e\rangle$ and $|g\rangle$, described by the Hamiltonian:
$$H = \hbar \omega |e\rangle\langle e|$$

* **Scenario:** If we prepare the system in state $|e\rangle$, there are no dynamics.
* **The Reality:** Physically, we expect the system to decay to $|g\rangle$ eventually because any transition has a finite line width (i.e., a finite lifetime).
* **The Reason:** The system is not isolated; it is coupled to a bath or environment (e.g., modes of the electromagnetic field).

### System-Environment Coupling
To describe this properly, we must include the environment:
* We consider the full system $S+E$, where $|\psi\rangle = |\psi_{SE}\rangle$.
* Transitions such as $|e\rangle \rightarrow |g\rangle$ and $|n_k\rangle \rightarrow |n_k+1\rangle$ become possible.
* **Challenge:** This is complicated due to the many degrees of freedom in the environment.

---

## Goal
Derive a description of system dynamics that accounts for dissipative effects (coupling to the environment) without needing to explicitly consider the environment's internal dynamics.

* **Desired Mapping:** We want a linear map $V(t)$ such that:
    $$\rho_S(t) = V(t) \rho_S(0)$$
    where $\rho_S = \text{tr}_E |\psi_{SE}\rangle\langle\psi_{SE}|$
* **The Picture:** $S$ and $E$ interact $\rightarrow$ they become entangled $\rightarrow$ the reduced density matrix $\rho_S$ becomes mixed $\rightarrow$ this results in non-unitary dynamics.
* **Physical Constraints:** The properties of $\rho_S$ must be preserved (Hermitian, positive, $\text{tr}(\rho_S)=1$). This leads to the requirement of a **Completely Positive Trace-Preserving (CPTP) map**.

| Feature | Unitary Operation | Non-Unitary Operation (CPTP) |
| :--- | :--- | :--- |
| **System Type** | Closed (Isolated) | Open (Coupled to Environment) |
| **Reversibility** | Fully reversible | Generally irreversible |
| **Information** | Preserved | Decays/Lost to the environment |
| **Math Tool** | Unitary Matrix ($U$) | Superoperator / Kraus Operators |
| **Entropy** | Stays constant | Usually increases |

## 2) Lindblad master equation
(Most general equation for $\rho_S$ fulfilling the requirements above)

**Note:** In classical physics, "master equation" is often used for an equation describing the evolution of a classical probability distribution (population dynamics). We refer to a **quantum master eq.** for populations and coherences of $S$. We call an eq. for the populations only, a **rate equation**.

Lindblad master equation: System state $\rho$ (with $\rho(t=0)=\rho_0$) evolves according to:

$$\dot{\rho} = -i [H, \rho] + \sum_{\alpha} \left( \Gamma_{\alpha} \rho \Gamma_{\alpha}^{\dagger} - \frac{1}{2} (\Gamma_{\alpha}^{\dagger} \Gamma_{\alpha} \rho + \rho \Gamma_{\alpha}^{\dagger} \Gamma_{\alpha}) \right)$$
*   **Von Neumann (unitary)**: $-i [H, \rho]$
*   **Lindblad terms $\mathcal{L}_{\alpha}[\rho]$ (dissipative)**: $\sum_{\alpha} \dots$

*   $\Gamma_{\alpha}$: jump operators
*   $\mathcal{L}[\rho]$: superoperators (map operators $\rho \to \mathcal{L}[\rho]$)


### Example (from intro)
($\hbar=1$), $H = \hbar \omega |e\rangle\langle e|$, $\rho_0 = |e\rangle\langle e|$, $\Gamma_{e \to g} = \sqrt{\gamma} |g\rangle\langle e|$

$$\dot{\rho} = -i (\omega |e\rangle\langle e|\rho - \omega \rho |e\rangle\langle e|) + \gamma \left[ |g\rangle\langle e|\rho |e\rangle\langle g| - \frac{1}{2} (|e\rangle\langle e|\rho + \rho |e\rangle\langle e|) \right]$$

*   $\dot{\rho}_{gg} = \langle g|\dot{\rho}|g\rangle = \gamma \langle e|\rho|e\rangle = \gamma \rho_{ee}$
*   $\dot{\rho}_{ee} = \langle e|\dot{\rho}|e\rangle = -i\omega (\rho_{ee} - \rho_{ee}) - \gamma \rho_{ee} \Rightarrow \rho_{ee}(t) = e^{-\gamma t}$
*   $\rho_{gg}(t) = 1 - e^{-\gamma t}$
*   $\dot{\rho}_{ge} = -\frac{\gamma}{2} \rho_{ge} + i\omega \rho_{ge}$ (stays zero)
*   ($\dot{\rho}_{gg} = \gamma e^{-\gamma t} \Rightarrow \rho_{gg} = \int_0^t \gamma e^{-\gamma t'} dt' = 1 - e^{-\gamma t}$)

*   $\rightarrow$ Population decay from $|e\rangle$ to $|g\rangle$ exponentially.
*   $\rightarrow$ Coherences decay exponentially.
*   $\rightarrow$ Trace & positivity conserved.

## 3) Deriving the Lindblad master equation
*(not covered in lect, optional, only note red underlined parts)*

In some cases the Lindblad ME can be derived rigorously. I show the main steps of the classic quantum optics way of deriving it.

*   $H = H_S + H_E + H_{SE}$
*   $H_{SE} = \sum_{\alpha} A_{\alpha} \otimes B_{\alpha}$
    *   $A_{\alpha}$ acting on $S$
    *   $B_{\alpha}$ acting on $E$

Full system $(S+E)$:
$$\dot{\rho}_{SE} = -i [H, \rho_{SE}]$$

Go to interaction picture: $\rho_{SE}^{(I)} = e^{i(H_S+H_E)t} \rho_{SE} e^{-i(H_S+H_E)t}$
$$\Rightarrow \dot{\rho}_{SE}^{(I)} = -i [H_{SE}^{(I)}, \rho_{SE}^{(I)}] \quad (1)$$
with $H_{SE}^{(I)} = e^{i(H_S+H_E)t} H_{SE} e^{-i(H_S+H_E)t}$

Formal solution of (1) (dropping indices $(I)$ and $SE$):
$$\rho = \rho(0) - i \int_0^t d\tau [H_{SE}(\tau), \rho(\tau)]$$

Since we want to get an equation for $\rho_S = \text{tr}_E(\rho_{SE})$ we insert this into (1) and take the **trace over the environment**:
$$\dot{\rho}_S = \text{tr}_E(\dot{\rho}) = -i \text{tr}_E([H_{SE}(t), \rho(0)]) - \int_0^t d\tau \text{tr}_E([H_{SE}(t), [H_{SE}(\tau), \rho(\tau)]])$$

Still exact, but not closed eq. for $\rho_S$. RHS depends on full environment dynamics.
$\Rightarrow$ Need to apply approximations!

* **Born approximation:** $\rho(t) \sim \rho_S(t) \otimes \rho_E$ ($+\mathcal{O}(\lambda)$)
    * $\hat{=}$ perturbative treatment of $H_{SE} \sim \lambda$
    * $\rho_E$ is constant (e.g. thermal equilibrium state)
    * $\hat{=}$ large reservoir assumption
    * also assume $\text{tr}\{B_{\alpha}(t)\rho_E\} = 0$ (without loss of generality)

* **1st Markov approximation:** $\rho_S(\tau) \sim \rho_S(t)$
    * $\hat{=}$ reservoir has no memory, bath correlations decay fast, eg. spont. emitted photons are lost
    * also assume $[H_{SE}, \rho(0)] = 0$ (without loss of gen.)

$$\Rightarrow \dot{\rho}_S = - \int_0^t d\tau \text{tr}_E \{[H_{SE}(t), [H_{SE}(\tau), \rho_S(t) \otimes \rho_E]]\}$$
**Born-Redfield master equation**

Closed equation for $\rho_S$ with time-dep. coeffs + preserves trace & hermiticity.

**Problems:**
* **time-dep. coeffs** $\rightarrow$ apply **2nd Markov approx**:
    * Put upper integration bound $t$ to $\infty$.
    * Justified by fast decaying bath-correlations $C_{\alpha\beta}(t_2-t_1) = \langle B_{\alpha}(t_1)B_{\beta}(t_2) \rangle$.
    * $\rightarrow$ large $t$'s don't contribute anyway.
* **Positivity not conserved**
    * $\rightarrow$ need more approximations: **Secular approx** (averaging over fast osc. terms) or **rotating wave approx**.
    * $\rightarrow$ Obtain Lindblad form, $\Gamma$ can be calculated from the bath correlations $C_{\alpha\beta}$ (problem dep.).

### Typical examples:
* Dynamics of electronic excitations in a solid coupled to phonon modes of the crystal (bath).
* Coupling of atoms to modes of the electromagnetic field (quantum optics) $\rightarrow$ Wigner-Weisskopf theory of spontaneous emission.

## 4) Solving the master equation / Properties of the Lindbladian
* Cast Lindblad ME into standard 1st order ODE system form (flatten $\rho$ into a vector):
  $$\dot{\rho} = \mathcal{L} \rho$$
  * $\mathcal{L}$: Liouvillian (Lindblad superop.) as matrix
  * $\rho = \sum_{mn} \rho_{mn} |m\rangle\langle n|$: vectorized density mat. (dim $d^2$; dim $2^{2N}$ for spin $1/2$)

* **Formal solution:** $\rho(t) = e^{\mathcal{L}t} \rho(0)$ ($\hat{=}$ CPTP map)
* $\mathcal{L}$ is not Hermitian $\rightarrow$ has complex eigenvalues.
* Trace preservation $\Rightarrow \text{Re}(\lambda) \le 0$ for all eigenvalues (only decay of populations).
* There is (at least) one state with $\mathcal{L}\rho_{ss} = 0$, called the **steady state** since $\rho(t) = e^{\mathcal{L}t} \rho(0) \xrightarrow{t \to \infty} \rho_{ss}$ for all $\rho(0)$.
* $\min (|\text{Re}(\lambda)|, \text{Re}(\lambda) < 0) =$ **Liouvillian gap** ($\rightarrow$ timescale on which $\rho_{ss}$ is approached).
* Closing of the Liouvillian gap $\rightarrow$ "dissipative phase transition", properties of $\rho_{ss}$ change.
