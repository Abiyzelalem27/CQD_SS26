## 1. `jax.numpy as jnp`

`jax.numpy` is similar to normal NumPy, but it works with **JAX**.

It is used for numerical calculations, especially in scientific computing, machine learning, and computational quantum dynamics.

In the Jastrow code, we use:

```python
sigma_z = 2 * s - 1
```

This converts spin values from `0/1` into physical spin values `-1/+1`.

We also use:

```python
jnp.sum(...)
jnp.roll(...)
```

These functions come from `jax.numpy`.

---

## Applications of `jax.numpy`

`jax.numpy` is used for:

* Linear algebra
* Matrix calculations
* Quantum state calculations
* Spin-chain simulations
* Optimization
* Automatic differentiation
* GPU/TPU acceleration
* Machine learning
* Monte Carlo simulations

In computational quantum dynamics, `jax.numpy` is useful because quantum states, Hamiltonians, and operators are usually represented using arrays and matrices.

For example:

```python
jnp.sum(sigma_z * jnp.roll(sigma_z, -1))
```

This computes nearest-neighbour spin correlations:

$$
\sum_i \sigma_i^z \sigma_{i+1}^z
$$

So, `jnp` performs the mathematical array calculations.

`flax.linen` is a library used to build trainable models in JAX.

In the code:

```python
class Jastrow(nn.Module):
```

the `Jastrow` class becomes a Flax model.

This means the Jastrow ansatz can have trainable parameters.

The symbol `nn` is used for:

```python
nn.Module
nn.compact
nn.initializers.zeros
```

## Applications of `flax.linen`

`flax.linen` is used for:

* Neural networks
* Trainable quantum ansatz models
* Variational wavefunctions
* Machine learning models
* Deep learning
* Neural quantum states

* Model optimization

In this exercise, `flax.linen` is used to create trainable parameters:

```python
J1 = self.param("J1", nn.initializers.zeros, ())
J2 = self.param("J2", nn.initializers.zeros, ())
```

## Simple Difference

```text
jax.numpy / jnp  = mathematical calculations with arrays

flax.linen / nn  = building trainable models with parameters
```

For the Jastrow code:

```text
jnp calculates the spin correlations.

nn creates the trainable Jastrow model.
```

Together,

$$
\texttt{jnp} + \texttt{nn}

## Place 2: Ground-State Search with SGD

In this section, we perform a ground-state search using stochastic gradient descent.

The objective is to minimize the variational energy

$$
E_\theta =
\frac{\langle \psi_\theta | H | \psi_\theta \rangle}
{\langle \psi_\theta | \psi_\theta \rangle}.
$$

The parameters of the variational wavefunction are updated using gradients of the energy.

Main goals:

* Define the local energy
* Estimate the variational energy using Monte Carlo samples
* Compute gradients with respect to the variational parameters
* Update parameters using `optax`
* Plot the variational energy versus iteration
* Compare the result with the exact ground-state energy
